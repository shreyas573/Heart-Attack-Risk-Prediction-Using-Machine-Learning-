```python
import pandas as pd
import numpy as np
import shap

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# =====================================================
# LOGISTIC REGRESSION MODEL
# =====================================================

best_model = lreg

feature_names = list(x_train.columns)
# =====================================================
# SHAP EXPLAINER (USE ALL TRAINING SAMPLES)
# =====================================================

masker = shap.maskers.Independent(
    x_train,
    max_samples=230
)

explainer = shap.Explainer(
    best_model,
    masker
)

# =====================================================
# INPUT FIELDS
# =====================================================

inputs = {}

for col in feature_names:

    inputs[col] = widgets.FloatText(
        description=col,
        layout=widgets.Layout(width='400px')
    )

predict_btn = widgets.Button(
    description='Predict Heart Risk',
    button_style='danger'
)

dashboard_output = widgets.Output()

# =====================================================
# PREDICTION FUNCTION
# =====================================================

def predict_patient(b):

    with dashboard_output:

        clear_output()

        # =====================================
        # USER INPUT
        # =====================================

        values = [[inputs[c].value for c in feature_names]]

        patient = pd.DataFrame(
            values,
            columns=feature_names
        )

        # =====================================
        # MODEL PREDICTION
        # =====================================

        pred = best_model.predict(patient)[0]

        try:
            prob = best_model.predict_proba(patient)[0][1] * 100

        except:
            prob = 50.0

        # =====================================
        # SAVE USER DATA TO CSV
        # =====================================

        record = patient.copy()

        record["Prediction"] = (
            "High Risk" if pred == 1 else "Low Risk"
        )

        record["Risk_Probability"] = round(prob, 2)

        record["Timestamp"] = pd.Timestamp.now()

        csv_file = r"C:\Users\HP\OneDrive\Documents\shreyas_project\patient_predictions.csv"

        try:

            existing_data = pd.read_csv(csv_file)

            updated_data = pd.concat(
                [existing_data, record],
                ignore_index=True
            )

            updated_data.to_csv(
                csv_file,
                index=False
            )

        except FileNotFoundError:

            record.to_csv(
                csv_file,
                index=False
            )

        display(HTML("""
        <div style="
        background:#0984e3;
        color:white;
        padding:10px;
        border-radius:10px;
        margin-bottom:15px">

        💾 Patient Data Saved Successfully

        </div>
        """))

        # =====================================
        # ALERT CARD
        # =====================================

        if pred == 1:

            display(HTML(f"""
            <div style="
            background:#d63031;
            color:white;
            padding:20px;
            border-radius:15px">

            <h2>🚨 HIGH HEART ATTACK RISK</h2>

            <h3>
            Risk Probability:
            {prob:.2f}%
            </h3>

            </div>
            """))

        else:

            display(HTML(f"""
            <div style="
            background:#27ae60;
            color:white;
            padding:20px;
            border-radius:15px">

            <h2>✅ LOW HEART ATTACK RISK</h2>

            <h3>
            Risk Probability:
            {prob:.2f}%
            </h3>

            </div>
            """))

        # =====================================
        # RISK METER
        # =====================================

        display(HTML(f"""
        <br>

        <h3>Risk Meter</h3>

        <progress
        value='{prob:.2f}'
        max='100'
        style='width:700px'>
        </progress>

        <h4>{prob:.2f}%</h4>
        """))

        # =====================================
        # SHAP XAI ANALYSIS
        # =====================================

        display(
            HTML(
                "<hr><h2>🔍 XAI Analysis</h2>"
            )
        )

        shap_values = explainer(patient)

        contrib = shap_values.values[0]

        xai_df = pd.DataFrame({

            "Feature": feature_names,

            "Value": patient.iloc[0],

            "Contribution": contrib

        })

        xai_df["Abs"] = np.abs(
            xai_df["Contribution"]
        )

        xai_df = xai_df.sort_values(
            "Abs",
            ascending=False
        )

        display(
            HTML(
                "<h3>Top Feature Contributions</h3>"
            )
        )

        display(
            xai_df[
                [
                    "Feature",
                    "Value",
                    "Contribution"
                ]
            ].head(10)
        )

        # =====================================
        # RISK FACTORS
        # =====================================

        if pred == 1:

            display(
                HTML(
                    "<h3>⚠ Factors Increasing Risk</h3>"
                )
            )

            risk_df = xai_df[
                xai_df["Contribution"] > 0
            ]

            display(
                risk_df[
                    [
                        "Feature",
                        "Value",
                        "Contribution"
                    ]
                ]
            )

            display(HTML("""

            <h3>
            ❤️ Recommended Safety Measures
            </h3>

            <ul>

            <li>Consult a Cardiologist</li>

            <li>Monitor Blood Pressure Regularly</li>

            <li>Reduce Cholesterol Levels</li>

            <li>Exercise 30 Minutes Daily</li>

            <li>Avoid Smoking</li>

            <li>Limit Alcohol Consumption</li>

            <li>Manage Stress Effectively</li>

            <li>Take Prescribed Medications</li>

            <li>Schedule Regular Heart Checkups</li>

            </ul>

            """))

        else:

            display(HTML("""

            <h3>
            ✅ Healthy Lifestyle Recommendations
            </h3>

            <ul>

            <li>Continue Regular Exercise</li>

            <li>Maintain Healthy Weight</li>

            <li>Follow Balanced Diet</li>

            <li>Avoid Smoking</li>

            <li>Get Annual Health Checkups</li>

            <li>Keep Cholesterol Under Control</li>

            <li>Maintain Healthy Blood Pressure</li>

            </ul>

            """))

        # =====================================
        # SHAP WATERFALL PLOT
        # =====================================

        print("\nGenerating SHAP Waterfall Plot...")

        shap.plots.waterfall(
            shap_values[0],
            max_display=13
        )

# =====================================================
# BUTTON EVENT
# =====================================================

predict_btn.on_click(
    predict_patient
)

# =====================================================
# DISPLAY DASHBOARD
# =====================================================

display(

    widgets.VBox(

        list(inputs.values())

        +

        [
            predict_btn,
            dashboard_output
        ]

    )

)
```
